Runnable companion to PDF Book II "How a dict really works".

A dict/set gives O(1) average lookup by turning a key into an array index via a
hash function. Two keys can map to the same slot — a COLLISION — and how you
resolve it is the whole game. This file builds BOTH classic strategies and
proves they behave like Python's dict:

  * SEPARATE CHAINING   — each slot holds a small list ("bucket") of entries
  * OPEN ADDRESSING     — on collision, probe forward to the next free slot
  * LOAD FACTOR & RESIZE— grow the table when it gets too full to keep O(1)

The lesson: O(1) is AMORTIZED and AVERAGE — a bad hash or a full table degrades
it to O(n). Resizing keeps the average low.

## 🧠 Mental Model: Hash Tables

> **Think of a hash table as a post-office with numbered boxes.**  
> The hash function is the address calculator: it converts any key into a box number in O(1). Two letters to the same box = collision — how you handle it determines performance.

### WHY — Why does it exist?
Arrays give O(1) access by *index*, but only if you know the index. Hash tables give O(1) access by *any key* — string, tuple, object — by computing an index from the key. This is the single most impactful data structure in practical software.

### WHAT — What is it?
An array of buckets + a hash function that maps keys → bucket indices. The hash function must be:
- **Deterministic**: same key → same index always
- **Uniform**: keys spread evenly across buckets (minimise collisions)
- **Fast**: O(1) to compute

### HOW — How does it work?

```
Key "name"  ──hash()──►  int 8765432  ──mod capacity──►  index 7
                                                              │
                                                    bucket[7] = [("name", "Ada")]
```

**Collision resolution strategies:**
| Strategy | Mechanism | Python's choice |
|----------|-----------|-----------------|
| Separate chaining | Each bucket is a list; colliding keys share the list | Historical dict (before 3.6) |
| Open addressing | Probe to next free slot; one flat array | CPython's current `dict` |

**Load factor** = `n_items / capacity`. When it exceeds ~0.75, resize (double the array, rehash everything). This keeps average chain length O(1) — the amortised magic.

### WHEN — When to use it?
- **O(1) average lookup/insert/delete** by any key: always prefer over a list for membership tests
- **Frequency counting**: `Counter`, `defaultdict(int)`
- **Memoisation cache**: function-result cache keyed by arguments
- **Grouping**: `defaultdict(list)` for grouping records

**Gotcha** — O(1) is *average*. Worst case (all keys collide) is O(n). A poor hash function or adversarial keys can degrade to O(n). Always use built-in `dict`/`set` in production — they use a battle-tested hash with randomisation (`PYTHONHASHSEED`).

```
Complexity:
  Average:  insert O(1)  lookup O(1)  delete O(1)
  Worst:    O(n) — when all keys hash to the same bucket
  Space:    O(n)  where n = number of stored keys
```
